Key Findings
1. Sales
The store generated ₵43,300 across 15 orders, with an average order value of ₵2,886.67.

2. Order Fulfillment
10 of the 15 orders were delivered, representing a delivery rate of 66.67%.

3. Payment Methods
Card and mobile money were the most frequently used payment methods, each accounting for 46.67% of orders.

4. Geographic Performance
Accra generated the highest number of orders, while Kumasi generated almost the same amount of revenue from substantially fewer orders.

5. Category Performance
Laptops generated the highest category-level sales value, while accessories recorded the highest unit sales.

6. Customer Engagement
4 out of 15 customers had not placed an order, representing 26.67% of the customer base.

7. Customer Feedback
Several reviewed products had high average ratings, although the number of reviews was relatively small.


-- Metric Validation Observation
During the analysis, a discrepancy was identified between order-level revenue and revenue calculated from order items.
The `orders` table reported total revenue of ₵43,300, while aggregating `quantity × unit_price` from order items produced a different total.
This highlighted the importance of defining metrics consistently before using them in a dashboard.
The order-level `total_amount` was therefore treated as the official order revenue metric, while line-item calculations were treated separately for product and category analysis.
This validation prevented different definitions of revenue from being presented as the same.
